```python
from pathlib import Path
from time import perf_counter

import nibabel as nib

dataset_path = Path("../data/raw/nifti_test_studies")

start = perf_counter()

files = [
    path
    for path in dataset_path.rglob("*")
    if path.is_file()
]

elapsed = perf_counter() - start

print(f"Archivos encontrados: {len(files)}")
print(f"Tiempo de descubrimiento: {elapsed:.4f} segundos")
```

```python
nifti_files = [
    path
    for path in files
    if path.name.endswith(".nii")
    or path.name.endswith(".nii.gz")
]

print(f"NIfTI encontrados: {len(nifti_files)}")
for path in nifti_files:
    print(path)
```

```python
loaded_images = []
error_count = 0

for path in nifti_files:
    try:
        image = nib.load(path)
        loaded_images.append((path, image))

    except (nib.filebasedimages.ImageFileError, OSError) as exc:
        error_count += 1
        print(f"Error: {path.name}")
        print(f"  {exc}")

print(f"\nNIfTI encontrados: {len(nifti_files)}")
print(f"NIfTI cargados: {len(loaded_images)}")
print(f"NIfTI con errores: {error_count}")
```

```python
for path, image in loaded_images:
    header = image.header

    print(f"\nArchivo: {path.name}")
    print(f"Shape: {image.shape}")
    print(f"Voxel size: {header.get_zooms()[:3]}")
    print(f"Orientation: {nib.aff2axcodes(image.affine)}")
    print(f"qform code: {int(header["qform_code"])}")
    print(f"sform code: {int(header["sform_code"])}")
```

```python
for path, image in loaded_images:
    volume = image.get_fdata()

    print(f"\nArchivo: {path.name}")
    print(f"Data type: {type(volume)}")
    print(f"Shape: {volume.shape}")
    print(f"Dtype: {volume.dtype}")
```

```python
start = perf_counter()

volumes = []

for path, image in loaded_images:
    volume = image.get_fdata()
    volumes.append((path, volume))

volume_time = perf_counter() - start

print(f"Tiempo de get_fdata(): {volume_time:.4f} segundos")
```
