```python
import nibabel as nib
from pathlib import Path

print(nib.__version__)
```

```python
nifti_path = Path(
    "../data/raw/nifti_test_studies/sub-001_ses-01_T1w.nii.gz"
    )

img = nib.load(nifti_path)
```

```python
print(type(img))
header = img.header

print(header)
```

```python
print(img.shape)
print(header.get_zooms())
print(header.get_data_dtype())
```

```python
data = img.get_fdata()

print(type(data))
print(data.shape)
print(data.dtype)

print(img.affine)
```

```python
print("Tipo:", type(img))
print("Shape:", img.shape)
print("Header:", type(img.header))
print("Voxel size:", img.header.get_zooms())
print("Datatype:", img.header.get_data_dtype())
print("Affine:")
print(img.affine)

print("Data type:", type(data))
print("Data shape:", data.shape)
print("Data dtype:", data.dtype)
```

```python
print("qform:")
print(img.header.get_qform())

print("\nsform:")
print(img.header.get_sform())

print("\nqform_code:", img.header["qform_code"])
print("sform_code:", img.header["sform_code"])
```

```python
for field in [
    "dim",
    "pixdim",
    "datatype",
    "bitpix",
    "scl_slope",
    "scl_inter",
]:
    print(field, "=", img.header[field])
```

```python
print("Min:", data.min())
print("Max:", data.max())
print("Mean:", data.mean())
```

```python
print(nib.aff2axcodes(img.affine))
print(nib.orientations.io_orientation(img.affine))
```
